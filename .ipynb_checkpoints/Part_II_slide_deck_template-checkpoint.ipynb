## Investigation Overview

This presentation conveys four key insights drawn from the exploratory 
analysis of 14 years of U.S. construction spending data (2002–2016):

1. Total private construction spending fell nearly 45% from peak to trough 
during the 2008 recession, with recovery still incomplete by 2016
2. Residential crashed immediately while office spending peaked during the 
recession — sectors follow fundamentally different cycles
3. Manufacturing overtook Commercial as the largest investment sector by 
2013–2016, representing a complete reversal of pre-recession hierarchy
4. Public spending acted as a countercyclical stabilizer, rising as private 
investment retreated during the recession

## Dataset Overview

The dataset contains monthly U.S. construction spending figures from January 
2002 to December 2016 (169 observations), sourced from the U.S. Census Bureau. 
Spending is broken down across 17 private and 17 public construction sectors, 
reported as annualized rates in millions of dollars. All values in this 
presentation have been converted to billions for readability.

```python
# Import all packages and set plots to be embedded inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb

%matplotlib inline

# Suppress warnings from final output
import warnings
warnings.simplefilter("ignore")

# Display settings for polished plots
plt.style.use('seaborn-v0_8-whitegrid')
sb.set_palette('muted')
pd.set_option('display.float_format', '{:.2f}'.format)
print('All packages loaded successfully.')
```

```python
# Load dataset and recreate era column used throughout the presentation
df = pd.read_csv('data/construction_spending.csv')

era_order = [
    'Pre-Recession (2002-2007)',
    'Recession (2008-2009)',
    'Early Recovery (2010-2012)',
    'Growth (2013-2016)'
]

def assign_era(year):
    """Classify a year into one of four economic eras."""
    if year <= 2007:
        return 'Pre-Recession (2002-2007)'
    elif year <= 2009:
        return 'Recession (2008-2009)'
    elif year <= 2012:
        return 'Early Recovery (2010-2012)'
    else:
        return 'Growth (2013-2016)'

df['era'] = df['time.year'].apply(assign_era)
print(f'Dataset loaded: {df.shape[0]} rows, {df.shape[1]} columns')
```

## Slide 1 - The Boom, The Crash, and The Recovery

Private construction spending tells a dramatic story of boom, collapse, 
and slow recovery across 14 years. Total private investment peaked at 
909B in 2006, then fell nearly 45% to a trough of 500.8B in 2011 — 
two full years after the official recession ended, confirming that 
construction recovery lags the broader economy. By 2016, spending had 
climbed back to 830B but had not yet reached its pre-recession peak.

```python
# Slide 1: Polished time series — Total Private Construction Spending Over Time

annual_private = df.groupby('time.year')['annual.private.total construction'].mean() / 1000

fig, ax = plt.subplots(figsize=(12, 6))

ax.plot(annual_private.index, annual_private.values,
        color='steelblue', linewidth=3, marker='o', markersize=6)

# Shade recession period
ax.axvspan(2008, 2009, alpha=0.15, color='red', label='Recession (2008–2009)')

# Shade eras with subtle fills
ax.axvspan(2002, 2007, alpha=0.05, color='steelblue')
ax.axvspan(2010, 2012, alpha=0.05, color='orange')
ax.axvspan(2013, 2016, alpha=0.05, color='green')

# Annotate peak and trough
peak_year = annual_private.idxmax()
trough_year = annual_private.idxmin()
ax.annotate(f'Peak: {annual_private[peak_year]:.0f}B (2006)',
            xy=(peak_year, annual_private[peak_year]),
            xytext=(peak_year - 2.5, annual_private[peak_year] + 20),
            arrowprops=dict(arrowstyle='->', color='black', lw=1.5),
            fontsize=11, fontweight='bold')
ax.annotate(f'Trough: {annual_private[trough_year]:.0f}B (2011)',
            xy=(trough_year, annual_private[trough_year]),
            xytext=(trough_year + 0.5, annual_private[trough_year] + 30),
            arrowprops=dict(arrowstyle='->', color='black', lw=1.5),
            fontsize=11, fontweight='bold')

# Add era labels
ax.text(2004, 530, 'Pre-Recession\nBoom', ha='center', fontsize=9,
        color='steelblue', fontweight='bold')
ax.text(2011.2, 530, 'Early\nRecovery', ha='center', fontsize=9,
        color='darkorange', fontweight='bold')
ax.text(2014.5, 530, 'Growth', ha='center', fontsize=9,
        color='seagreen', fontweight='bold')

ax.set_title('Total Private Construction Spending (2002–2016)',
             fontsize=15, fontweight='bold', pad=15)
ax.set_xlabel('Year', fontsize=12)
ax.set_ylabel('Average Annualized Spending Rate (Billions of Dollars)', fontsize=12)
ax.set_xticks(annual_private.index)
ax.set_xticklabels(annual_private.index, rotation=45)
ax.legend(fontsize=11)
ax.set_ylim(450, 980)

plt.tight_layout()
plt.show()
```

## Slide 2 - Not All Sectors Fall Equally

When the recession hit, residential construction collapsed immediately — 
falling 60% from its 630B peak to 250B by 2009. Office spending told a 
completely different story, actually peaking during the recession as 
developers completed already-committed projects. This lagged response 
reveals a critical insight: office investment is a lagging indicator, 
giving developers a 2-3 year warning window if they track residential 
as an early signal.

```python
# Slide 2: Office spending peaked DURING recession while residential crashed immediately

fig, axes = plt.subplots(2, 1, figsize=(12, 9))

# Annual means in billions
annual_office = df.groupby('time.year')['annual.private.office'].mean() / 1000
annual_residential = df.groupby('time.year')['annual.private.residential'].mean() / 1000

# Top panel — Office
axes[0].plot(annual_office.index, annual_office.values,
             color='steelblue', linewidth=2.5, marker='o', markersize=5)
axes[0].axvspan(2008, 2009, alpha=0.2, color='red', label='Recession (2008–2009)')
axes[0].annotate('Office peaks DURING recession',
                 xy=(2008, annual_office[2008]),
                 xytext=(2003, annual_office[2008] - 8),
                 arrowprops=dict(arrowstyle='->', color='black', lw=1.5),
                 fontsize=10, fontweight='bold', color='firebrick')
axes[0].set_title('Private Office Construction Spending (2002–2016)',
                  fontsize=13, fontweight='bold')
axes[0].set_ylabel('Avg Spending (Billions)', fontsize=11)
axes[0].set_xticks(annual_office.index)
axes[0].set_xticklabels(annual_office.index, rotation=45, fontsize=8)
axes[0].legend(fontsize=10)

# Bottom panel — Residential
axes[1].plot(annual_residential.index, annual_residential.values,
             color='firebrick', linewidth=2.5, marker='o', markersize=5)
axes[1].axvspan(2008, 2009, alpha=0.2, color='red', label='Recession (2008–2009)')
axes[1].annotate('Residential crashes BEFORE recession band',
                 xy=(2007, annual_residential[2007]),
                 xytext=(2010.5, annual_residential[2007] + 50),
                 arrowprops=dict(arrowstyle='->', color='black', lw=1.5),
                 fontsize=10, fontweight='bold', color='firebrick')
axes[1].set_title('Private Residential Construction Spending (2002–2016)',
                  fontsize=13, fontweight='bold')
axes[1].set_ylabel('Avg Spending (Billions)', fontsize=11)
axes[1].set_xlabel('Year', fontsize=11)
axes[1].set_xticks(annual_residential.index)
axes[1].set_xticklabels(annual_residential.index, rotation=45, fontsize=8)
axes[1].legend(fontsize=10)

fig.suptitle('Office vs. Residential: Two Very Different Recession Stories',
             fontsize=14, fontweight='bold', y=1.01)

plt.tight_layout()
plt.show()
```

## Slide 3 - The Sector Hierarchy Reversal

Before the recession, Commercial real estate dominated private construction 
investment. By the Growth era (2013–2016), Manufacturing had overtaken 
Commercial as the largest sector — a complete reversal of the pre-recession 
hierarchy. This shift signals that post-recession private capital flowed 
into industrial infrastructure rather than traditional commercial real estate, 
fundamentally changing the landscape for developers.

```python
# Slide 3: Polished clustered bar — sector hierarchy reversal across eras

sectors_s3 = {
    'annual.private.office': 'Office',
    'annual.private.commercial': 'Commercial',
    'annual.private.manufacturing': 'Manufacturing',
    'annual.private.lodging': 'Lodging'
}

era_sector_means = {}
for col, label in sectors_s3.items():
    era_sector_means[label] = [
        df[df['era'] == e][col].mean() / 1000 for e in era_order
    ]

x = np.arange(len(era_order))
width = 0.2
colors_s3 = ['steelblue', 'darkorange', 'seagreen', 'firebrick']

fig, ax = plt.subplots(figsize=(14, 7))

for i, (label, values) in enumerate(era_sector_means.items()):
    bars = ax.bar(x + i * width, values, width,
                  label=label, color=colors_s3[i], alpha=0.88)
    for bar in bars:
        ax.text(bar.get_x() + bar.get_width()/2,
                bar.get_height() + 0.8,
                f'{bar.get_height():.1f}B',
                ha='center', fontsize=8.5, fontweight='bold')

# Highlight the reversal with an annotation
ax.annotate('Manufacturing overtakes\nCommercial in Growth era',
            xy=(3 + width, era_sector_means['Manufacturing'][3]),
            xytext=(2.2, 85),
            arrowprops=dict(arrowstyle='->', color='black', lw=1.5),
            fontsize=10, fontweight='bold', color='seagreen')

ax.set_title('Private Sector Investment Hierarchy Across Economic Eras (2002–2016)',
             fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Economic Era', fontsize=12)
ax.set_ylabel('Average Annualized Spending Rate (Billions of Dollars)', fontsize=12)
ax.set_xticks(x + width * 1.5)
ax.set_xticklabels([e.replace(' (', '\n(') for e in era_order], fontsize=10)
ax.legend(fontsize=10)
ax.set_ylim(0, 110)

plt.tight_layout()
plt.show()
```

## Slide 4 - The Public Backstop

When private developers retreated during the recession, government 
construction investment did the opposite — rising from 237.9B to 311.1B 
while private spending fell sharply. This countercyclical relationship 
narrowed the private-to-public spending ratio from 3.3:1 pre-recession 
to just 1.8:1 during the Early Recovery era. For developers, understanding 
this dynamic is critical — public infrastructure investment partially 
sustained the construction workforce and supply chain during the years 
when private capital was scarce.

```python
# Slide 4: Polished clustered bar — Private vs. Public spending by era

private_by_era = df.groupby('era')['annual.private.total construction'].mean() / 1000
public_by_era = df.groupby('era')['annual.public.total construction'].mean() / 1000

x = np.arange(len(era_order))
width = 0.35

fig, ax = plt.subplots(figsize=(12, 7))

bars1 = ax.bar(x - width/2, [private_by_era[e] for e in era_order],
               width, label='Private', color='steelblue', alpha=0.88)
bars2 = ax.bar(x + width/2, [public_by_era[e] for e in era_order],
               width, label='Public', color='firebrick', alpha=0.88)

# Annotate values
for bar in bars1:
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 5,
            f'{bar.get_height():.1f}B', ha='center',
            fontsize=10, fontweight='bold', color='steelblue')
for bar in bars2:
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 5,
            f'{bar.get_height():.1f}B', ha='center',
            fontsize=10, fontweight='bold', color='firebrick')

# Add ratio annotations
ratios = [private_by_era[e]/public_by_era[e] for e in era_order]
for i, (ratio, era) in enumerate(zip(ratios, era_order)):
    ax.text(i, 20, f'Ratio\n{ratio:.1f}:1',
            ha='center', fontsize=9, fontweight='bold',
            color='black',
            bbox=dict(boxstyle='round,pad=0.3', facecolor='lightyellow',
                      edgecolor='gray', alpha=0.8))

ax.annotate('Public spending rises\nas private retreats',
            xy=(1 + width/2, public_by_era['Recession (2008-2009)']),
            xytext=(2, public_by_era['Recession (2008-2009)'] + 80),
            arrowprops=dict(arrowstyle='->', color='black', lw=1.5),
            fontsize=10, fontweight='bold', color='firebrick')

ax.set_title('Private vs. Public Construction Spending by Economic Era',
             fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Economic Era', fontsize=12)
ax.set_ylabel('Average Annualized Spending Rate (Billions of Dollars)', fontsize=12)
ax.set_xticks(x)
ax.set_xticklabels([e.replace(' (', '\n(') for e in era_order], fontsize=10)
ax.legend(fontsize=11)
ax.set_ylim(0, 1000)

plt.tight_layout()
plt.show()
```
